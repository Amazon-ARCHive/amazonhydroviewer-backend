```python
hydrobasins_lev05_url = 'https://raw.githubusercontent.com/blackteacatsu/spring_2024_envs_research_amazon_ldas/main/resources/hybas_sa_lev05_areaofstudy.geojson'

with urllib.request.urlopen(hydrobasins_lev05_url) as url:
    geojson_data = json.loads(url.read().decode())
    # Combine all polygons into one trace
    polygon_x = []  # Store all x-coordinates
    polygon_y = []  # Store all y-coordinates
    polygon_text = []  # Store hover text (PFAF_ID)
    
    for feature in geojson_data['features']:
        pfaf_id = feature['properties']['PFAF_ID']
        if feature['geometry']['type'] == 'Polygon':
            x, y = zip(*feature['geometry']['coordinates'][0])
            polygon_x.extend(x + (None,))  # Add None to break the line between polygons
            polygon_y.extend(y + (None,))
            polygon_text.extend([pfaf_id] * len(x) + [None])  # Add None for breaks
        elif feature['geometry']['type'] == 'MultiPolygon':
            for polygon in feature['geometry']['coordinates']:
                x, y = zip(*polygon[0])
                polygon_x.extend(x + (None,))
                polygon_y.extend(y + (None,))
                polygon_text.extend([pfaf_id] * len(x) + [None])


    # Add a single trace for all polygons
    # heatmap=go.FigureWidget(
    #     go.Scatter(
    #     x=polygon_x,
    #     y=polygon_y,
    #     mode='lines',
    #     line=dict(color='white', width=1.5),
    #     text=polygon_text,
    #     hoverinfo='text',
    #     hovertemplate='PFAF_ID: %{text}<extra></extra>'
    #     )
    # )
```

```python
fig = go.Figure(
    go.Scattermap(
        lon=polygon_x,
        lat=polygon_y,
        mode='lines',
        line=dict(color='grey', width=1.5),
        text=polygon_text,
        hoverinfo='text',
        hovertemplate='PFAF_ID: %{text}<extra></extra>'
    )
)
# fig.update_layout(map_style="satellite", 
#                   map_center={'lat' : -7.5, 'lon' : -65.5},
#                   map_zoom = 3.5,
#                   margin={"r":0,"t":0,"l":0,"b":0})
```

```python
# Add tile server layer to Plotly Maplibre
import requests

# Configuration
TILE_SERVER_URL = "http://localhost:5000"
variable = "Rainf_tavg"
time_idx = 0
category = 2
profile = 0
vmin = 40  # Minimum value for colormap
vmax = 100  # Maximum value for colormap

# Clear cache first to ensure fresh tiles
try:
    requests.get(f"{TILE_SERVER_URL}/cache/clear")
    print("✓ Cache cleared")
except:
    print("⚠ Could not clear cache")

# Build tile URL template with mode=global, tms=false, and vmin/vmax
# Plotly uses XYZ coordinates (Y=0 at north), so tms=false
tile_url = f"{TILE_SERVER_URL}/tiles/{variable}/{time_idx}/{category}/{{z}}/{{x}}/{{y}}.png?colormap=Reds&profile={profile}&mode=global&tms=true&vmin={vmin}&vmax={vmax}"

# Data bounds from pyramid
data_bounds = {'lon_min': -81.975, 'lon_max': -49.025, 'lat_min': -20.975, 'lat_max': 5.975}

# Configure map with custom raster layer
# Note: Plotly zoom levels are global Web Mercator
# Your pyramid zoom 4 (grain=30, coarsest) should now be available
fig.update_layout(
    map=dict(
        style="carto-voyager",
        center=dict(lat=-7.5, lon=-65.5),
        zoom=3,  # Start at zoom 4 to see your coarsest data (grain=30)
        layers=[
            dict(
                sourcetype="raster",
                source=[tile_url],
                below="",  # Layer order
                opacity=0.7
            )
        ]
    ),
    margin={"r":0,"t":0,"l":0,"b":0}
)

fig.show()
```
